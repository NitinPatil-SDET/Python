Que1> Given two arrays a[] and b[], check if they are disjoint, i.e., there is no element common between both the arrays. Return true if if they are disjoint, otherwise, false.

```python
def areDisjoint(a,b):
    return len(set(a).intersection(set(b))) == 0

a = [1, 2, 3]
b = [4, 5, 6]

print(areDisjoint(a,b))
```

Que2> You are given two arrays a[] and b[], return the Union of both the arrays in any order.

The Union of two arrays is a collection of all distinct elements present in either of the arrays. If an element appears more than once in one or both arrays, it should be included only once in the result.

Note: Elements of a[] and b[] are not necessarily distinct.
Note that, You can return the Union in any order but the driver code will print the result in sorted order only.

```python
def findUnion(a,b):
    return set(a).union(set(b))

a = [1, 2, 3, 2, 1]
b = [3, 2, 2, 3, 3, 2]
print(findUnion(a,b))
```

Que3> Given two integer arrays a[] and b[], you have to find the intersection of the two arrays. Intersection of two arrays is said to be elements that are common in both the arrays. The intersection should not have duplicate elements and the result should contain items in any order.

Note: The driver code will sort the resulting array in increasing order before printing.

```python
def intersect(a,b):
    return set(a).intersection(set(b))


a = [1, 2, 3, 2, 1]
b = [3, 2, 2, 3, 3, 2]
print(intersect(a,b))
```

Que4> Given an array arr[] of integers and another integer target. Determine if there exist two distinct indices such that the sum of their elements is equal to the target.

```python
def twoSum(arr, target):
    seen = set()

    for num in arr:
        compliment = num - target

        if compliment in seen :
            return True
        seen.add(num)
    return False

arr = [0, -1, 2, -3, 1]
target = -2

print(twoSum(arr, target))
```
