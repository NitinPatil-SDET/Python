Que1> Given two arrays a[] and b[], check if they are disjoint, i.e., there is no element common between both the arrays. Return true if if they are disjoint, otherwise, false.

```python
def areDisjoint(a,b):
    return len(set(a).intersection(set(b))) == 0

a = [1, 2, 3]
b = [4, 5, 6]

print(areDisjoint(a,b))
```
