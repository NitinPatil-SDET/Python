Que1> Given two arrays a[] and b[], check if they are disjoint, i.e., there is no element common between both the arrays. Return true if if they are disjoint, otherwise, false.

```python
def areDisjoint(a,b):
    return len(set(a).intersection(set(b))) == 0

a = [1, 2, 3]
b = [4, 5, 6]

print(areDisjoint(a,b))
```

Que2> You are given two arrays a[] and b[], return the Union of both the arrays in any order.

The Union of two arrays is a collection of all distinct elements present in either of the arrays. If an element appears more than once in one or both arrays, it should be included only once in the result.

Note: Elements of a[] and b[] are not necessarily distinct.
Note that, You can return the Union in any order but the driver code will print the result in sorted order only.

```python
def findUnion(a,b):
    return set(a).union(set(b))

a = [1, 2, 3, 2, 1]
b = [3, 2, 2, 3, 3, 2]
print(findUnion(a,b))
```
