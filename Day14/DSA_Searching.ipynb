Que1> Given an array of positive integers arr[], return the second largest element from the array. If the second largest element doesn't exist then return -1.

Note: The second largest element should not be equal to the largest element.

Examples:

Input: arr[] = [12, 35, 1, 10, 34, 1]
Output: 34
Explanation: The largest element of the array is 35 and the second largest element is 34.

```python
def first_repeated(arr):
    seen = set()
    min_index = -1

    for i in range(len(arr)-1, -1, -1):
        if arr[i] in seen:
            min_index=i
        else :
            seen.add(arr[i])
    if min_index != -1:
        return min_index+1
    else:
        return -1

arr = [12, 35, 1, 10, 34, 1]
arr2 = [1, 5, 3, 4, 3, 5, 6]
print(first_repeated(arr))
print(first_repeated(arr2))
```
