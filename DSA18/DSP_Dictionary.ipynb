Que1> Given an array arr[], the task is to find the maximum distance between two occurrences of an element. If no element has two occurrences, then return 0.

```python
def maxDistance(arr):
    maxDistanc = {}
    res = 0
    for i in range(len(arr)):
        if arr[i] not in maxDistanc:
            maxDistanc[arr[i]]=i
        else:
            dist = i-maxDistanc[arr[i]]
            res = max(res, dist)
    return res

arr = [1, 1, 2, 2, 2, 1]
print(maxDistance(arr))

    
```
