Que1> You are given a string s, and your task is to reverse the string.

Examples:

Input: s = "Nitin"
Output: "nitiN"

```python
def reverseString(str):
    return str[::-1]

print(reverseString("Nitin"))
```

Que2> Given a non-empty sequence of characters s, return true if sequence is Binary, else return false.

Examples:

Input: s = "101"
Output: true
Explanation: Since string contains only '0' and '1', output is true.

```python
def isBinary(str):
    for ch in str:
        if ch != '0' and ch != '1':
            return False
    return True

print(isBinary("10101"))
print(isBinary("234"))
```

Que4> Given a string s, check if it is a "Panagram" or not. Return true if the string is a Panagram, else return false.
A "Panagram" is a sentence containing every letter in the English Alphabet either in lowercase or Uppercase.

```python
def checkPangram(s):
    s=s.lower()
    letter = set(s)
    for ch in "abcdefghijklmnopqrstuvwxyz":
        if ch not in letter:
            return False
    return True

s = "The quick brown fox jumps over the lazy dog"
print(checkPangram(s))
```
